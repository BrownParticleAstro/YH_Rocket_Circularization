## Introduction to Reinforcement learning

### Theoretical introduction

*What is Reinforcement learning ?*

Reinforcement Learning (RL) is a unique branch of machine learning that stands apart from other types, such as supervised, semi-supervised, and unsupervised learning. What makes RL distinct is its closed-loop structure and the way the model learns through interaction and feedback rather than direct instruction.  

In RL, the model—known as the **agent**—actively interacts with an **environment** to make decisions. It is not given explicit instructions about which actions to take. Instead, the agent learns through a process of trial and error, receiving feedback in the form of **rewards** or **penalties** based on the outcomes of its actions. The ultimate goal is to maximize cumulative rewards by gradually discovering the best strategy, or **policy**, for the task at hand.  

Unlike other machine learning methods, RL involves a dynamic feedback loop. Each action taken by the agent directly influences the environment, altering its **state**. This new state serves as the next input for the agent, shaping its subsequent decisions. This interdependent relationship between actions and outcomes is what defines the closed-loop nature of RL and makes it particularly well-suited for problems requiring sequential decision-making, such as controlling robots, optimizing systems, or navigating environments like games or simulations.  

By learning through interaction and adapting to changing states, RL enables models to autonomously solve complex problems where traditional static data-driven approaches fall short.


  
In order to understand RL and the game we're working on some basic terminology must be covered. To introduce those

*Some Basic Terms*

1. **State**  
   At every moment (or **timestep**), the spaceship exists in a specific condition defined by its **state**.  
   - Think of the **state** as a snapshot of all the relevant information in our case  the spaceship's position, velocity, and distance from the black hole.(but that is something that could be changed)
   - The state is what the agent "sees" and uses to decide the next step.

2. **Action**  
   To navigate space, the spaceship can apply **thrust** in different directions. Each possible adjustment to the spaceship’s movement—whether increasing or decreasing thrust—is an **action**.  
   - In this example, the agent can choose from actions like “increase thrust forward,” “apply reverse thrust,” or “remain idle.”

3. **Action Space**  
   The **action space** is the set of all possible actions the agent can take.  
   - For our spaceship, the action space might include thrust adjustments for now only tengentially to the trajectory.
   - There are different types of action spaces possible :
    - Discrete : the action space consists of finite set of distinct actions like. For example it would be if the agent only had the option to thrust forward, backward or stop applying thrust {-1,1,0}
    - Box : when the action is chosen between a continous range of value in this case in between -1 and 1 : [-1,1]
  

4. **State Space**  
   The **state space** refers to the collection of all possible states the spaceship can be in.  
   - For example, the state space could represent all combinations of positions and velocities relative to the black hole.  
   - RL algorithms often need to handle extremely large state spaces, especially in complex environments.
   - In our case the type of state space is a 4 dimensional box where the position and velocities can have continous values in a specific range.

5. **Agent**  
   The **agent** is the decision-maker. It’s the brain behind the spaceship's actions.  
   - The agent's goal is to figure out the best way to control the spaceship by learning from the feedback (rewards) it receives.

6. **Policy**  
   The **policy** is the agent’s strategy. It’s a function that maps a given state (like the spaceship’s current position and velocity) to the best action to take (like the right amount of thrust in the right direction).  
   - Over time, the agent improves its policy to make better decisions.
  
7. **Reward**  
   Each time the spaceship takes an action, the environment provides a **reward** as feedback.  
   - For example, if the spaceship gets closer to the desired orbit, the agent might receive a positive reward. Conversely, if the spaceship drifts too close to the black hole, the agent might get a negative reward.  
   - Rewards are the agent's guide, helping it learn what works and what doesn't.

8. **Value Function** \\
  The **value function** evaluates how good it is for the agent to be in a specifici state or take a specific action, given the expected future rewards.
  - V(s) usually refers to the state value function wich represents the total expected reard the agent can achieve from a particular state s following the current policy.
  - Q(s,a) usually refers to the action value function. It extends the state value function by also considering the sepcific action a evaluating the expected reward if it starts in state s and takes action a following the current policy.

Goal : *Understanding what it means*

When training a PPO, we can understand its behavior as following two phases. 

In the first one, the agent (or) is actually playing the game. No updates or optimization is happening, it only saves all the information in a buffer. The buffer will contain every states at each time step the associated action, probability distribution and rewards received.
In the second it uses the saved informations to evaluate the policy and both neural network : the actor and critic gets updated. 
It is in that second loop that the surregate loss will be computed, it is its gradient descent that will allow the optimization of the two networks.

Understanding the Advantage  : 

The advantage is what the model takes as the actual reflection of the greatness of the action taken. The principle is to compare the "value" that the Critic predicts the state the object was in had, and the reward the action taken by the actor had. To understand the importance of the advantage, an easy scenario can be taken into account. If the spaceship was in a really bad spot but took the right action, its reward will still be low because of its initial state. However, after trianing, we expect the critic to be able to detect that and give a lower value to that state, comparing the two will show that not a large update is necessary. 
 
Now we can also notice that we are not comparing the reward itsled and the value, we are comparing a discounted reward. That basically adds another piece of information, the duration of the episode after the action taken. For the same action, an episode that lasts longer will have a bettwe discounted reward than the one of a shortened episode. It kind of helps picturing how durable an action could be.

```python
# @title From Discreet to Continous Actions
'Changing the environment to get from discreet to continous actions'
"""
Classic cart-pole system implemented by Rich Sutton et al.
Copied from http://incompleteideas.net/sutton/book/code/pole.c
permalink: https://perma.cc/C9ZM-652R
Continuous version by Ian Danforth
"""

import math
import gym
from gym import spaces, logger
from gym.utils import seeding
import numpy as np

class ContinuousCartPoleEnv(gym.Env):
    metadata = {
        'render.modes': ['human', 'rgb_array'],
        'video.frames_per_second': 50
    }

    def __init__(self):
        self.gravity = 9.8
        self.masscart = 1.0
        self.masspole = 0.1
        self.total_mass = (self.masspole + self.masscart)
        self.length = 0.5  # actually half the pole's length
        self.polemass_length = (self.masspole * self.length)
        self.force_mag = 30.0
        self.tau = 0.02  # seconds between state updates
        self.min_action = -1.0
        self.max_action = 1.0

        # Angle at which to fail the episode
        self.theta_threshold_radians = 12 * 2 * math.pi / 360
        self.x_threshold = 2.4

        # Angle limit set to 2 * theta_threshold_radians so failing observation
        # is still within bounds
        high = np.array([
            self.x_threshold * 2,
            np.finfo(np.float32).max,
            self.theta_threshold_radians * 2,
            np.finfo(np.float32).max])

################# IMPORTANT SECTION ###################################
        self.action_space = spaces.Box(
            low=self.min_action,
            high=self.max_action,
            shape=(1,)
        )
#######################################################################
        self.observation_space = spaces.Box(-high, high)

        self.seed()
        self.viewer = None
        self.state = None

        self.steps_beyond_done = None

    def seed(self, seed=None):
        self.np_random, seed = seeding.np_random(seed)
        return [seed]

    def stepPhysics(self, force):
        x, x_dot, theta, theta_dot = self.state
        costheta = math.cos(theta)
        sintheta = math.sin(theta)
        temp = (force + self.polemass_length * theta_dot * theta_dot * sintheta) / self.total_mass
        thetaacc = (self.gravity * sintheta - costheta * temp) / \
            (self.length * (4.0/3.0 - self.masspole * costheta * costheta / self.total_mass))
        xacc = temp - self.polemass_length * thetaacc * costheta / self.total_mass
        x = x + self.tau * x_dot
        x_dot = x_dot + self.tau * xacc
        theta = theta + self.tau * theta_dot
        theta_dot = theta_dot + self.tau * thetaacc
        return (x, x_dot, theta, theta_dot)

    def step(self, action):
        assert self.action_space.contains(action), \
            "%r (%s) invalid" % (action, type(action))
        # Cast action to float to strip np trappings
        force = self.force_mag * float(action)
        self.state = self.stepPhysics(force)
        x, x_dot, theta, theta_dot = self.state
        done = x < -self.x_threshold \
            or x > self.x_threshold \
            or theta < -self.theta_threshold_radians \
            or theta > self.theta_threshold_radians
        done = bool(done)

        if not done:
            reward = 1.0
        elif self.steps_beyond_done is None:
            # Pole just fell!
            self.steps_beyond_done = 0
            reward = 1.0
        else:
            if self.steps_beyond_done == 0:
                logger.warn("""
You are calling 'step()' even though this environment has already returned
done = True. You should always call 'reset()' once you receive 'done = True'
Any further steps are undefined behavior.
                """)
            self.steps_beyond_done += 1
            reward = 0.0

        return np.array(self.state), reward, done, {}

    def reset(self):
        self.state = self.np_random.uniform(low=-0.05, high=0.05, size=(4,))
        self.steps_beyond_done = None
        return np.array(self.state)

    def render(self):
        if self.render_mode is None:
            gym.logger.warn(
                "You are calling render method without specifying any render mode. "
                "You can specify the render_mode at initialization, "
                f'e.g. gym("{self.spec.id}", render_mode="rgb_array")'
            )
            return

        try:
            import pygame
            from pygame import gfxdraw
        except ImportError:
            raise DependencyNotInstalled(
                "pygame is not installed, run `pip install gym[classic_control]`"
            )

        if self.screen is None:
            pygame.init()
            if self.render_mode == "human":
                pygame.display.init()
                self.screen = pygame.display.set_mode(
                    (self.screen_width, self.screen_height)
                )
            else:  # mode == "rgb_array"
                self.screen = pygame.Surface((self.screen_width, self.screen_height))
        if self.clock is None:
            self.clock = pygame.time.Clock()

        world_width = self.x_threshold * 2
        scale = self.screen_width / world_width
        polewidth = 10.0
        polelen = scale * (2 * self.length)
        cartwidth = 50.0
        cartheight = 30.0

        if self.state is None:
            return None

        x = self.state

        self.surf = pygame.Surface((self.screen_width, self.screen_height))
        self.surf.fill((255, 255, 255))

        l, r, t, b = -cartwidth / 2, cartwidth / 2, cartheight / 2, -cartheight / 2
        axleoffset = cartheight / 4.0
        cartx = x[0] * scale + self.screen_width / 2.0  # MIDDLE OF CART
        carty = 100  # TOP OF CART
        cart_coords = [(l, b), (l, t), (r, t), (r, b)]
        cart_coords = [(c[0] + cartx, c[1] + carty) for c in cart_coords]
        gfxdraw.aapolygon(self.surf, cart_coords, (0, 0, 0))
        gfxdraw.filled_polygon(self.surf, cart_coords, (0, 0, 0))

        l, r, t, b = (
            -polewidth / 2,
            polewidth / 2,
            polelen - polewidth / 2,
            -polewidth / 2,
        )

        pole_coords = []
        for coord in [(l, b), (l, t), (r, t), (r, b)]:
            coord = pygame.math.Vector2(coord).rotate_rad(-x[2])
            coord = (coord[0] + cartx, coord[1] + carty + axleoffset)
            pole_coords.append(coord)
        gfxdraw.aapolygon(self.surf, pole_coords, (202, 152, 101))
        gfxdraw.filled_polygon(self.surf, pole_coords, (202, 152, 101))

        gfxdraw.aacircle(
            self.surf,
            int(cartx),
            int(carty + axleoffset),
            int(polewidth / 2),
            (129, 132, 203),
        )
        gfxdraw.filled_circle(
            self.surf,
            int(cartx),
            int(carty + axleoffset),
            int(polewidth / 2),
            (129, 132, 203),
        )

        gfxdraw.hline(self.surf, 0, self.screen_width, carty, (0, 0, 0))

        self.surf = pygame.transform.flip(self.surf, False, True)
        self.screen.blit(self.surf, (0, 0))
        if self.render_mode == "human":
            pygame.event.pump()
            self.clock.tick(self.metadata["render_fps"])
            pygame.display.flip()

        elif self.render_mode == "rgb_array":
            return np.transpose(
                np.array(pygame.surfarray.pixels3d(self.screen)), axes=(1, 0, 2)
            )


    def close(self):
        if self.viewer:
            self.viewer.close()
```
